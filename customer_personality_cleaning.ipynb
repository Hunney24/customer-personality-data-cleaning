```python
import pandas as pd

df = pd.read_csv("marketing_campaign.csv", sep=";")

df.head()
```

```python
print("Shape of dataset:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("MISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())
```

```python
print("Income statistics:")
print(df["Income"].describe())

print("\nMissing Income records:")
print(df[df["Income"].isnull()])
```

```python
income_median = df["Income"].median()

print("Median Income:", income_median)
print("Missing Income before treatment:", df["Income"].isnull().sum())
```

```python

df["Income"] = df["Income"].fillna(income_median)

print("Missing Income after treatment:", df["Income"].isnull().sum())
```

```python
print("Education values:")
print(df["Education"].value_counts())

print("\nMarital Status values:")
print(df["Marital_Status"].value_counts())
```

```python
print(df[df["Marital_Status"].isin(["Alone", "Absurd", "YOLO"])]
      [["ID", "Year_Birth", "Education", "Marital_Status", "Income", "Kidhome", "Teenhome"]])
```

```python
df["Education"] = df["Education"].replace(
    "2n Cycle", "2nd Cycle"
)

print(df["Education"].value_counts())
```

```python
print("Data type:")
print(df["Dt_Customer"].dtype)

print("\nSample dates:")
print(df["Dt_Customer"].head(10))

print("\nUnique date formats/examples:")
print(df["Dt_Customer"].unique()[:20])
```

```python
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    errors="coerce"
)

print(df["Dt_Customer"].dtype)
```

```python
print("Missing/invalid dates after conversion:",
      df["Dt_Customer"].isnull().sum())
```

```python
print(df.dtypes)
```

```python
print("\nUnique values for categorical columns:")

for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
df.describe().T
```

```python
print("Data types:\n")
print(df.dtypes)
```

```python
print(df["Year_Birth"].describe())

print("\nUnique birth years:")
print(sorted(df["Year_Birth"].unique())[:20])
```

```python
old_birth_years = df[df["Year_Birth"] <= 1900]

print(old_birth_years[
    ["ID", "Year_Birth", "Education", "Marital_Status", "Income", "Dt_Customer"]
])
```

```python
reference_year = df["Dt_Customer"].dt.year.max()

df["Age"] = reference_year - df["Year_Birth"]

print(df[df["Year_Birth"] <= 1900][
    ["ID", "Year_Birth", "Age", "Income", "Dt_Customer"]
])
```

```python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    negative_count = (df[col] < 0).sum()

    if negative_count > 0:
        print(f"{col}: {negative_count} negative values")
```

```python
print("Total rows:", len(df))
print("Unique IDs:", df["ID"].nunique())
```

```python
binary_cols = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5",
    "Complain",
    "Response"
]

for col in binary_cols:
    print(col, ":", sorted(df[col].unique()))
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
print("FINAL DATA QUALITY CHECK")
print("-" * 40)

print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Unique IDs:", df["id"].nunique())
print("Invalid dates:", df["dt_customer"].isna().sum())

print("\nData types:")
print(df.dtypes)

print("\nEducation values:")
print(df["education"].value_counts())

print("\nMarital Status values:")
print(df["marital_status"].value_counts())
```
